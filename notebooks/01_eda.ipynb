```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
df = pd.read_csv(Path("C:\\Users\\91990\\Desktop\\ML\\Sentiment Analysis\\data\\Reviews.csv"))
print(df.head())
print(df.size)
print(df.shape)
print(df.columns)
```

```python
print("Missing values in each column")
print(df.isnull().sum())
print("--------------------------")
df_new=df.drop(["ProductId","UserId","Id"],axis=1)
print("Duplicate rows in the dataset: ", df_new.duplicated().sum())
print("--------------------------")
# print("Column data types:", df.dtypes)
# print("--------------------------")
print(df.duplicated(subset=["UserId","ProductId","Text"]).sum())
# print(df[df.duplicated(subset=["Score","Summary","Text"])])
print(df[df.duplicated(subset=["Text"])])

df=df.drop_duplicates(subset=["Text"])
print("DONE")
print(df.duplicated(subset=["Text"]).sum())

```

```python
print((df["Score"].value_counts(normalize=True).sort_index()*100).plot(kind="bar"))
plt.show()
```

```python
def label_sentiment(score):
    if score == 3:
        return "Neutral"
    elif score < 3:
        return "Negative"
    else:
        return "Positive"
    
df["Sentiment"]=df["Score"].apply(label_sentiment)

print((df["Sentiment"].value_counts(normalize=True).sort_index()*100).plot(kind="bar"))
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Percentage of reviews")
plt.show()
```

```python
#WE ONLY NEED TEXT AND SENTIMENT COLUMNS FOR OUR TASK
df=df[["Text","Sentiment"]]
df=df[df['Text'].str.strip()!=' ']

print(df.shape)
print(df.columns)

df.to_csv("../data/cleaned_reviews.csv", index=False)
```
