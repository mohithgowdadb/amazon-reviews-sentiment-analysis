#  Model Comparison

## 1. Objective

The objective of this phase is to evaluate and compare a traditional machine learning model, Logistic Regression, with a PyTorch Multilayer Perceptron (MLP) for three-class sentiment classification.

The model predicts one of three sentiment classes:
- Negative
- Neutral
- Positive

## 2. Evaluation Strategy

The dataset was divided into training, validation, and test sets.

- **Training set:** Used to train the models.
- **Validation set:** Used to compare models and assess their performance during experimentation.
- **Test set:** Used for the final evaluation of the selected model on unseen data.

The models were evaluated using accuracy, precision, recall, F1-score, and confusion matrices. Macro F1 and class-wise F1-scores are particularly important because the dataset has an imbalanced sentiment distribution.

## 3. Logistic Regression — Validation Results

Logistic Regression was used as the traditional machine learning baseline, with TF-IDF features representing the review text.

| Metric | Validation Score |
|---|---:|
| Accuracy | 0.86 |
| Macro F1 | 0.61 |
| Weighted F1 | 0.84 |
| Negative F1 | 0.69 |
| Neutral F1 | 0.22 |
| Positive F1 | 0.93 |

### Observations

- Logistic Regression achieved 86% validation accuracy.
- It performed well on Positive reviews.
- Neutral sentiment was the most challenging class, with an F1-score of 0.22.
- The model's high overall accuracy does not fully reflect its weaker performance on minority classes.

## 4. PyTorch MLP — Validation Results

A Multilayer Perceptron was implemented using PyTorch. TF-IDF vectors were used as input features, and the network was trained to classify reviews into the three sentiment categories.

The training process used a suitable loss function and the Adam optimizer.

| Metric | Validation Score |
|---|---:|
| Accuracy | 0.82 |
| Macro F1 | 0.63 |
| Weighted F1 | 0.83 |
| Negative F1 | 0.67 |
| Neutral F1 | 0.31 |
| Positive F1 | 0.91 |

### Observations

- The MLP achieved 82% validation accuracy.
- It achieved a slightly higher macro F1-score than Logistic Regression.
- Its Neutral F1-score improved from 0.22 to 0.31.
- Positive sentiment remained the easiest class to classify.
- Experiments with undersampling and different TF-IDF feature counts showed that changing the training distribution and feature representation affected performance.

## 5. Model Comparison

| Metric | Logistic Regression | PyTorch MLP |
|---|---:|---:|
| Validation Accuracy | 0.86 | 0.82 |
| Validation Macro F1 | 0.61 | 0.63 |
| Validation Weighted F1 | 0.84 | 0.83 |
| Negative F1 | 0.69 | 0.67 |
| Neutral F1 | 0.22 | 0.31 |
| Positive F1 | 0.93 | 0.91 |

### Comparison Summary

- Logistic Regression achieved higher validation accuracy and weighted F1.
- The MLP achieved a higher validation macro F1-score.
- The MLP performed better on Neutral reviews according to F1-score.
- Neither model consistently outperformed the other across all metrics.

These results demonstrate that a neural network is not automatically better than a traditional machine learning model. Model performance depends on the data representation, training strategy, and evaluation metric.

## 6. PyTorch MLP — Final Test Results

After the model experiments, the MLP was evaluated on the held-out test set.

| Metric | Test Score |
|---|---:|
| Accuracy | 0.77 |
| Macro F1 | 0.61 |
| Weighted F1 | 0.80 |
| Negative F1 | 0.64 |
| Neutral F1 | 0.30 |
| Positive F1 | 0.87 |

### Test Set Error Analysis

- **Negative sentiment:** The model achieved an F1-score of 0.64.
- **Neutral sentiment:** The model achieved an F1-score of 0.30, indicating continued difficulty distinguishing neutral reviews from the other sentiment classes.
- **Positive sentiment:** The model achieved an F1-score of 0.87, performing better than on the other classes.

The test accuracy and class-wise F1-scores were lower than the corresponding validation results. This indicates that the model's performance declined on unseen data and should be considered when assessing its generalization ability.

## 7. Limitations

- The dataset contains substantially more Positive reviews than Neutral and Negative reviews.
- Neutral sentiment can overlap linguistically with both Positive and Negative sentiment.
- TF-IDF represents text using word-based numerical features but does not fully capture context, sarcasm, or complex language.
- The MLP did not consistently outperform Logistic Regression.
- Further experiments with class weighting, alternative text representations, and model architectures may improve performance, but improvements are not guaranteed.

## 8. Conclusion

This phase compared Logistic Regression, a traditional machine learning baseline, with a PyTorch MLP for three-class sentiment analysis.

Logistic Regression achieved higher validation accuracy, while the MLP achieved a slightly higher validation macro F1-score and better Neutral-class F1-score. The MLP achieved 77% accuracy and a macro F1-score of 0.61 on the held-out test set.
